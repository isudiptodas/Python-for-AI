# Class & Objects

```python
class Dog:
  def __init__(self, name, sound):
    self.name = name
    self.sound = sound
  def getName(self):
    print(f"Dog name is {self.name}")
  def getSound(self):
    print(f"Dog sound is {self.sound}")

newDog = Dog(name='beagle', sound='bark')

newDog.getName()
newDog.getSound()
```

```python
class Bank:
  def __init__(self, bankName, accountNumber, balance = 0):
    self.bank_name = bankName
    self.account_number = accountNumber
    self.__balance = balance

  def getBalance(self):
    print(f"Available balance is : {self.__balance}\n")

  def deposit(self, money):
    self.__balance += money
    print(f"Your account has been credited with {money}\n")

  def withdraw(self, amount):
    if self.__balance < amount:
      print("Insufficient balance\n")
    else:
      self.__balance-=amount
      print(f"Amount {amount} has been debited from your account\n")

bank1 = Bank("cbi", "1234567")
bank2 = Bank("sbi", "2374738")

bank1.getBalance()

bank2.deposit(500)
bank2.deposit(1000)

bank1.getBalance()
bank2.getBalance()

bank2.withdraw(500)
bank2.getBalance()
```

```python
class Car:
  def __init__(self, type, version = 1):
    self.type = type
    self.version = version

class AstonMartin(Car):
  def __init__(self, year, model, type):
    self.year = year
    self.model = model
    version = 1
    super().__init__(type, version)

  def printType(self):
    print(f"Car type is {self.type}\n")

  def modifyVersion(self):
    self.version += 1
    print(f"{self.model} version modified\n")

  def printVersion(self):
    print(f"Current {self.model} version is {self.version}\n")

aston = AstonMartin(2022, "superleggera", "petrol")

aston.printType()
aston.printVersion()
aston.modifyVersion()
aston.modifyVersion()
aston.printVersion()
```
