```python

from youtube_transcript_api import YouTubeTranscriptApi
ytt_api = YouTubeTranscriptApi()

def fetchTranscript(URL):
  try:
    videoID = URL.split('/')[3].split('?')[0]

    transcript = ytt_api.fetch(videoID)
    return transcript

  except Exception as e:
    print(f"An error occured. Can't fetch transcripts")
    return None

URL = input("Enter youtube URL : ")

if "/shorts/" in URL:
  URL = input("Short form content is not allowed : ")

transcripts = fetchTranscript(URL)

if transcripts:
  for snippet in transcripts:
    print(f"{snippet.text}\n")
else:
  print("No transcripts found")
```
