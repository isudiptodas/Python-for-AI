```python
from youtube_transcript_api import YouTubeTranscriptApi
from google.colab import drive

drive.mount('/content/drive')
ytt_api = YouTubeTranscriptApi()

def fetchTranscript(URL):
  try:
    videoID = URL.split('/')[3].split('?')[0]

    transcript = ytt_api.fetch(videoID)
    addInFile(transcript, videoID)
    return transcript

  except Exception as e:
    print(f"An error occured. Can't fetch transcripts")
    return None

def addInFile(transcripts, id):
  try:
    if transcripts:
      filepath = f"/content/drive/MyDrive/colab_files/{id}.txt"
      with open(filepath, 'w') as file:
        for snippet in transcripts:
          file.write(f"{snippet.text}\n")
      print("Transcripts saved in file")
  except Exception as e:
    print(f"File saving error\n {e}")

URL = input("Enter youtube URL : ")

if "/shorts/" in URL:
  URL = input("Short form content is not allowed : ")

transcripts = fetchTranscript(URL)

if transcripts:
  print("Transcripts fetched successfully")
else:
  print("No transcripts found")
```
