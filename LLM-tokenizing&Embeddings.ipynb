# Tokenizing

```python

import re

# class for tokenizing
class Tokenizer:
  def __init__(self, vocab_list):
    self.token_to_id = {token:id for id, token in enumerate(vocab_list)}
    self.id_to_token = {id:token for id, token in enumerate(vocab_list)}

  # function to convert words to numbers
  def encode(self, text):
    processed = [word for word in re.findall(r"[A-Za-z0-9]+|[^A-Za-z0-9\s]", text) if word and not word.isspace()]
    encoded = []

    for word in processed:
      if word in self.token_to_id:
        encoded.append(self.token_to_id[word])
      else:
        encoded.append(self.token_to_id["<|UNK|>"])

    return encoded

  # function to convert numbers to words
  def decode(self, id):
    words = []
    for i in id:
      words.append(self.id_to_token[i])

    return " ".join(words)

text = ""

with open("/content/drive/MyDrive/colab_files/english_data.txt", "r", encoding="utf-8") as f:
  text = f.read()

# word level tokenizing
words = [word for word in re.findall(r"[A-Za-z0-9]+|[^A-Za-z0-9\s]", text.strip()) if word and not word.isspace()]

# character level tokenizing
chars = []
for word in text:
  if word.strip():
    chars.append(word)

all_tokens = sorted(list(set(words)))
all_tokens.extend(["<|unk|>"])

vocab_size = len(all_tokens)
vocab_list = {token:id for id, token in enumerate(all_tokens)}

tokenizer = Tokenizer(vocab_list)
print(tokenizer.encode("Who is Alaska young mr. brian ?"))
print(tokenizer.decode([1469, 4260, 108, 7353, 7383, 6, 7383, 69]))
```
